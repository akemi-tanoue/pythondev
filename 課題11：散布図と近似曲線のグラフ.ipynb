```python
import matplotlib.pyplot as plt # グラフや図を描くライブラリを呼び出す
import numpy as np # 数値計算や分析をするライブラリを呼び出す

# データ
x = np.array([62, 95, 83, 100, 72, 71, 55, 85, 67, 75])   # english
y = np.array([62, 80, 85, 95, 58, 76, 52, 84, 67, 71])    # math

# 散布図（点は青色で、ラベルはscore）
plt.scatter(x, y, color="blue", label="score")

# 近似直線（最小二乗法を利用する、直線は赤色で、ラベルはlsm）
a, b = np.polyfit(x, y, 1) # 最小二乗法の近似直線を計算してくれる関数、1は1次式（直線）の意味
y_fit = a * x + b #直線の式
plt.plot(x, y_fit, color="red", label="lsm") # 近似直線上のyの値を求めて描写する

# 横軸と縦軸それぞれのラベルを表示
plt.xlabel("english") # 横軸はenglish
plt.ylabel("math") # 縦軸はmath

# 凡例 位置は左上とする
plt.legend(loc="upper left")

plt.show()

# 課題11：散布図と近似曲線のグラフ.ipynb というノートブックを作成します
# 以下のコードを貼り付けます。
# import numpy as np

# x = np.array([62, 95, 83, 100, 72, 71, 55, 85, 67, 75])   # english
# y = np.array([62, 80, 85, 95, 58, 76, 52, 84, 67, 71])    # math

# 上記の x と y のndarrayを利用してください　→○
# このデータの散布図グラフを描いてください：　→○
# 横軸が x 、縦軸は y としてください　→○
# 散布図の点は青色にしてください　→○
# 最小二乗法を利用して、このデータの近似直線を散布図と一緒に描いてください：　→○
# 直線は赤色にしてください　→○
# 横軸と縦軸それぞれのラベルを表示してください：　→○
# 横軸のラベルは english としてください　→○
# 縦軸のラベルは math としてください　→○
# 凡例を表示してください：　→○
# 表示する位置は、左上としてください　→○
# 表示する文字列は、散布図のほうは score、近似直線のほうは lsm としてください　→○
```
