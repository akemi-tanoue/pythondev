```python
# 最初のセルに Baseball_Team という名前のクラスを作ってください：
# Baseball_Teamに、以下のメンバ変数を持たせてください：
# name（チーム名が入ります：Archers／Meteors／Cheetahs／Sharks／Pythons／Owls）
# win（勝ち数）
# lose（負け数）
# draw（引き分け数）
class Baseball_Team: #Baseball_Teamというクラスを作る
  def __init__(self, name, win, lose, draw): #name,win,lose,drawを変数として持たせる
    self.name = name
    self.win = win
    self.lose = lose
    self.draw = draw
# Baseball_Teamに、 勝率を算出する calc_win_rate() というメソッドを作ってください：
# 勝率は、勝ち数/(勝ち数+負け数) にて算出します
# 今回は勝率をメンバ変数で持つ形ではなく、メソッドによって算出する形とします。勝率の値が欲しいときは、このcalc_win_rate()メソッドを実行します
# なお、今回は勝率の算出において、引き分け数は考慮しない形とします
  def calc_win_rate(self): #calc_win_rate() というメソッドを作る
    total_games = self.win + self.lose #試合数=(勝ち数+負け数)
    return self.win / total_games #勝率＝勝ち数÷試合数で計算して、その結果を返す
# Baseball_Teamに、勝敗情報を表示する show_team_result() というメソッドを作ってください：
# 勝敗情報の表示は、下記「出力結果の例」の表の1行分を表示するメソッドにしてください。
  def show_team_result(self): #一行で表示する
    print(
      f"{self.name:<9} {self.win:3d} {self.lose:4d} {self.draw:4d} {self.calc_win_rate():.3f}" #出力内容の整形をする
    )
# 例：Archers 77 64 2 0.546
# 次のセルで Baseball_Team のインスタンスを6個チーム分、生成してください：
# これらのインスタンスには、それぞれのチームの名前、勝ち数、負け数、引き分け数を格納します。
teams = [
  Baseball_Team("Archers", 77, 64, 2),
  Baseball_Team("Meteors", 71, 69, 3),
  Baseball_Team("Cheetahs", 69, 68, 6),
  Baseball_Team("Sharks", 70, 70, 3),
  Baseball_Team("Pythons", 68, 73, 2),
  Baseball_Team("Owls", 59, 82, 2)
]
# 各インスタンスの show_team_result() メソッドを呼び出して、勝敗情報を表示してください：
# 「出力結果の例」のように、表の見出しも表示してください。
print(f"{'team':<9} {'win':>3} {'lose':>4} {'draw':>4} {'rate':<6}") #タイトルの設定と整形を出力内容に合わせる
for team in teams: #それぞれ出力する
  team.show_team_result()

# 補足情報
# 書式指定を活用して、きれいな表を出力できていれば高評価としますが、任意とします。対応しないと不合格になるものではありません
# 書式指定について詳しくは公式ドキュメントを参照してください：
# 書式指定文字列の文法 - Python 3.8 ドキュメント
# 継承、多態性、カプセル化への対応も任意とします
```
